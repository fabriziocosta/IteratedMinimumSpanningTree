# Interactive 2D IMSTE

Run the code cell below in Jupyter to load the sample pool and create the live controls. Select MNIST, Fashion-MNIST, or Kuzushiji-MNIST; MNIST is loaded first and other pools load when selected, then stay cached. Defaults are 1,000 samples, 15 MSTs with FAMST, squared Euclidean distance, log attraction, logistic repulsion, rank-scheduled updates, and 200 epochs (adjustable from 50 to 2,000). Distance is always squared Euclidean, attraction is logarithmic, and repulsion is logistic. Logistic margin and temperature control repulsion. The sliders allow up to 64 MSTs, a FAMST maximum of 512 neighbors (default 120); FAMST targets two neighbors per requested MST, subject to its maximum, and a negative ratio of 30. Fits use the selected epoch count and a 0.05 learning rate, with separate progress bars for MST extraction and embedding optimization. The repulsion-share slider ranges from 0 to 1 and defaults to 0.5. Choose **MST builder** to compare approximate FAMST (default) with exact Prim. FAMST requires `python -m pip install -e ".[famst]"`. The graph and negative sampling run on CPU. Choose MPS in the device menu to run batched coordinate updates on an available Apple GPU; MPS can produce a different update trajectory. Click **Fit embedding** to run. The optional five-fold 5-NN estimate is off by default. Use **Estimator** to choose transductive IMSTE (training rows only) or inductive IMSTE (MLP projection). For the inductive estimator, choose the number of hidden layers, units per layer, and dropout; defaults are 6 layers, 128 units, and 0.1 dropout. Training uses a 10% validation split, at least 100 epochs, and patience of 20. The inductive estimator uses its MLP for training and novel rows alike. The training slider can use up to 60,000 MNIST rows, reserving the remaining rows for novel predictions. Click **Compare reference and novel predictions** to compare the exact optimized embedding with MLP predictions for disjoint novel rows, up to the training sample count. **Reset sliders** restores the defaults.
