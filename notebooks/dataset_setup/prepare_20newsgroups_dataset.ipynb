# Prepare 20 Newsgroups vectors

Fetch and clean 20 Newsgroups, then prepare features with either Google EmbeddingGemma 2 or TF-IDF. Gemma defaults to four documents per batch and a 2,048-token limit to keep memory use manageable. Labels are saved for coloring in the [visualization notebook](../20newsgroups_exp.ipynb).

Install once with `%pip install -e ".[notebook,preprocessing]"`. Gemma 2 downloads model weights on its first run. TF-IDF uses a progress bar for batched vectorization and reduces the sparse vectors to dense features for the same visualization workflow. The settings cell currently uses 50 messages per category for a quick run; set `MAX_PER_CATEGORY = None` to use the full dataset.

```python
from pathlib import Path
import sys

REPO_ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents)
                 if (path / "pyproject.toml").is_file() and (path / "imste").is_dir())
SRC = str(REPO_ROOT / "notebooks" / "src")
if SRC not in sys.path:
    sys.path.insert(0, SRC)

import importlib
import newsgroups
newsgroups = importlib.reload(newsgroups)
prepare_20newsgroups_dataset = newsgroups.prepare_20newsgroups_dataset

DATA_DIR = REPO_ROOT / "data" / "20newsgroups"
EMBEDDING_METHOD = "gemma2"  # Choose: "gemma2" or "tfidf".
TFIDF_DIMENSIONS = 256  # Used only when EMBEDDING_METHOD is "tfidf".
MAX_PER_CATEGORY = None  # Set to None for all documents, or a positive cap per category.
RANDOM_STATE = 42
BATCH_SIZE = 4
MAX_SEQ_LENGTH = 2048  # Gemma 2 truncates longer messages to this length.
FORCE_REEMBED = False
```

```python
X, texts, labels, indices, category_names, manifest = prepare_20newsgroups_dataset(
    DATA_DIR,
    embedding_method=EMBEDDING_METHOD,
    tfidf_dimensions=TFIDF_DIMENSIONS,
    max_per_category=MAX_PER_CATEGORY,
    random_state=RANDOM_STATE,
    batch_size=BATCH_SIZE,
    max_seq_length=MAX_SEQ_LENGTH,
    force_reembed=FORCE_REEMBED,
)
```
